```python
seats = ["A1", "A2", "A3", "A4", "A5"]

print("Available Seats:")
print(seats)

seat = input("Enter seat number to book: ")

if seat in seats:
    seats.remove(seat)
    print("Seat", seat, "booked successfully.")
else:
    print("Seat is not available.")

print("Remaining Seats:")
print(seats)
```
